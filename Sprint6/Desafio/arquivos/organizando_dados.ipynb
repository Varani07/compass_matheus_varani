# Configurações Iniciais

## Realizando Imports Necessários

```python
import pandas as pd
from dotenv import load_dotenv
import boto3
```

## Lendo Dataset

- Utilizando `usecols` para filtrar apenas as colunas que acredito serem importantes para as próximas etapas.

```python
df = pd.read_csv('comprasGOV_original.csv', usecols=['ano_compra', 'descricao_resumida', 'quantidade', 'unidade_medida', 'valor_unitario_estimado', 'valor_total', 'valor_unitario_resultado', 'valor_total_resultado', 'situacao_compra_item_nome', 'tem_resultado'])
```

## Gerando Novo Arquivo CSV

```python
df.to_csv('comprasGOV_modificado.csv', index=False)
```

# Subindo Arquivo Para um Bucket no Amazon S3

## Carregando variáveis de ambiente

```python
load_dotenv()
```

## Salvando arquivo no Bucket

- `boto3.resource('s3')` cria um ponto de acesso para interagir com o serviço Amazon S3.
- `s3.Bucket('desafio-sprint-06')` gera um objeto para representar o bucket que desejamos gerenciar.
- `bucket.upload_file('comprasGOV_modificado.csv', 'comprasGOV.csv')` envia o arquivo para o bucket referenciado pelo objeto, o primeiro argumento simboliza o arquivo local, o segundo é como o arquivo será nomeado dentro do bucket.

```python
s3 = boto3.resource('s3')
bucket = s3.Bucket('desafio-sprint-06')
bucket.upload_file('comprasGOV_modificado.csv', 'comprasGOV.csv')
```
